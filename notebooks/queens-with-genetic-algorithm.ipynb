```python
# Define the grid
grid = [
    [1, 2, 2, 2, 2],
    [3, 4, 3, 2, 3],
    [3, 4, 3, 3, 3],
    [3, 3, 3, 5, 5],
    [3, 3, 3, 3, 3],
]

colors = {1, 2, 3, 4, 5}  # Unique color numbers

# Parameters for GA
POPULATION_SIZE = 100
NUM_GENERATIONS = 500
MUTATION_RATE = 0.1

def fitness(solution):
    penalties = 0
    
    # Row and column constraints
    columns = set()
    for row, col in enumerate(solution):
        if col in columns:
            penalties += 1  # Column conflict
        columns.add(col)
    
    # Color constraints
    color_set = set()
    for row, col in enumerate(solution):
        color = grid[row][col]
        if color in color_set:
            penalties += 1  # Color conflict
        color_set.add(color)
    
    # Proximity constraints
    for i, (row1, col1) in enumerate(enumerate(solution)):
        for row2, col2 in enumerate(solution):
            if i == row2:
                continue
            if abs(row1 - row2) <= 1 and abs(col1 - col2) <= 1:
                penalties += 1  # Proximity conflict
    
    return penalties

def generate_population(size):
    population = []
    for _ in range(size):
        individual = [random.randint(0, len(grid[0]) - 1) for _ in range(len(grid))]
        population.append(individual)
    return population

def crossover(parent1, parent2):
    point = random.randint(1, len(parent1) - 2)
    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]
    return child1, child2

def mutate(individual):
    if random.random() < MUTATION_RATE:
        row = random.randint(0, len(individual) - 1)
        individual[row] = random.randint(0, len(grid[0]) - 1)

def genetic_algorithm():
    population = generate_population(POPULATION_SIZE)
    best_solution = None
    best_fitness = float('inf')
    
    for generation in range(NUM_GENERATIONS):
        # Evaluate fitness
        fitness_scores = [fitness(individual) for individual in population]
        
        # best_idx = np.argmin(fitness_scores)
        best_idx = -1
        for i in range (len(fitness_scores)):
            if fitness_scores[i] < best_fitness:
                best_idx = i
                best_fitness = fitness_scores[best_idx]
                best_solution = population[best_idx]
        
        # Update the best solution
        if fitness_scores[best_idx] < best_fitness:
            best_fitness = fitness_scores[best_idx]
            best_solution = population[best_idx]
        
        if best_fitness == 0:
            break  # Solution found
        
        # Selection
        selected = random.choices(population, weights=[1/f for f in fitness_scores], k=POPULATION_SIZE)
        
        # Crossover
        next_generation = []
        for i in range(0, POPULATION_SIZE, 2):
            parent1, parent2 = selected[i], selected[i + 1]
            child1, child2 = crossover(parent1, parent2)
            next_generation.extend([child1, child2])
        
        # Mutation
        for individual in next_generation:
            mutate(individual)
        
        population = next_generation
    
    return best_solution, best_fitness

# Run the genetic algorithm
solution, fitness_score = genetic_algorithm()

print("Best solution:", solution)
print("Fitness score:", fitness_score)
```
