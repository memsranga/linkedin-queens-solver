```python
# Constants for the Genetic Algorithm (GA)
POPULATION_SIZE = 100  # Number of individuals in the population
NUM_GENERATIONS = 500  # Number of generations to evolve
MUTATION_RATE = 0.1    # Probability of mutation for an individual

# Grid representing the problem's constraints (colors)
grid = [
    [1, 2, 2, 2, 2],
    [3, 4, 3, 2, 3],
    [3, 4, 3, 3, 3],
    [3, 3, 3, 5, 5],
    [3, 3, 3, 3, 3],
]

# Set of unique colors in the grid
colors = {1, 2, 3, 4, 5}

# Fitness function to evaluate the quality of a solution
def fitness(solution):
    """
    Calculate the fitness score of a solution.
    Lower fitness scores are better.
    Penalties are incurred for:
    - Column conflicts (two rows with the same column index).
    - Color conflicts (two rows with the same color value).
    - Proximity conflicts (adjacent cells row-wise, column-wise, or diagonally).
    """
    penalties = 0
    
    # Check for column conflicts
    columns = set()
    for row, col in enumerate(solution):
        if col in columns:
            penalties += 1
        columns.add(col)
    
    # Check for color conflicts
    color_set = set()
    for row, col in enumerate(solution):
        color = grid[row][col]
        if color in color_set:
            penalties += 1
        color_set.add(color)
    
    # Check for proximity conflicts
    for i, (row1, col1) in enumerate(enumerate(solution)):
        for row2, col2 in enumerate(solution):
            if i == row2:
                continue
            if abs(row1 - row2) <= 1 and abs(col1 - col2) <= 1:
                penalties += 1  # Conflict if cells are adjacent
    
    return penalties

# Function to generate the initial population
def generate_population(size):
    """
    Generate a population of random individuals.
    Each individual is a list where the index represents the row, and the value represents the column.
    """
    population = []
    for _ in range(size):
        individual = [random.randint(0, len(grid[0]) - 1) for _ in range(len(grid))]
        population.append(individual)
    return population

# Function to perform crossover between two parents
def crossover(parent1, parent2):
    """
    Perform single-point crossover to generate two children.
    """
    point = random.randint(1, len(parent1) - 2)  # Select a random crossover point
    child1 = parent1[:point] + parent2[point:]  # First child combines parent1 and parent2
    child2 = parent2[:point] + parent1[point:]  # Second child combines parent2 and parent1
    return child1, child2

# Function to mutate an individual
def mutate(individual):
    """
    Mutate an individual with a probability defined by MUTATION_RATE.
    Mutation changes the column index of a random row.
    """
    if random.random() < MUTATION_RATE:
        row = random.randint(0, len(individual) - 1)  # Choose a random row
        individual[row] = random.randint(0, len(grid[0]) - 1)  # Assign a new random column

# Genetic Algorithm main function
def genetic_algorithm():
    """
    Run the Genetic Algorithm to find the best solution.
    """
    # Initialize the population
    population = generate_population(POPULATION_SIZE)
    best_solution = None
    best_fitness = float('inf')  # Initialize with a very high fitness score
    
    # Iterate through generations
    for generation in range(NUM_GENERATIONS):
        # Evaluate fitness for each individual
        fitness_scores = [fitness(individual) for individual in population]
        
        # Update the best solution if a better one is found
        best_idx = fitness_scores.index(min(fitness_scores))
        if fitness_scores[best_idx] < best_fitness:
            best_fitness = fitness_scores[best_idx]
            best_solution = population[best_idx]
        
        # Exit early if a perfect solution is found
        if best_fitness == 0:
            break
        
        # Selection: Weighted random sampling based on fitness (lower fitness is better)
        selected = random.choices(population, weights=[1 / f for f in fitness_scores], k=POPULATION_SIZE)
        
        # Crossover: Generate the next generation
        next_generation = []
        for i in range(0, POPULATION_SIZE, 2):
            parent1, parent2 = selected[i], selected[i + 1]
            child1, child2 = crossover(parent1, parent2)
            next_generation.extend([child1, child2])
        
        # Mutation: Mutate individuals in the next generation
        for individual in next_generation:
            mutate(individual)
        
        # Replace the old population with the new one
        population = next_generation
        print(f"After {generation} generation, best fitness is: {best_fitness} and solution is: {best_solution}")
    
    return best_solution, best_fitness

# Run the Genetic Algorithm and print the results
solution, fitness_score = genetic_algorithm()

print("Best solution:", solution)
print("Fitness score:", fitness_score)
```
